# **Eksplorasi Data (EDA)**

## **Comparison (Perbandingan)**

**1. Tingkat Stres Berdasarkan Platform Media Sosial**

```python
stress_platform = df.groupby('platform_usage')['stress_level'].mean().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot( x=stress_platform.index, y=stress_platform.values,palette='Blues')
plt.title('Rata-rata Tingkat Stres Berdasarkan Platform Media Sosial')
plt.xlabel('Platform')
plt.ylabel('Rata-rata Stress Level')
plt.show()
```

Berdasarkan visualisasi bar chart rata-rata tingkat stres berdasarkan platform media sosial, terlihat bahwa pengguna Instagram, TikTok, maupun pengguna kedua platform (Both) memiliki rata-rata tingkat stres yang hampir sama, yaitu berada pada kisaran nilai 5 hingga 6.

Perbedaan rata-rata tingkat stres antar platform terlihat sangat kecil sehingga tidak menunjukkan adanya pengaruh yang signifikan dari jenis platform media sosial yang digunakan terhadap tingkat stres remaja. Hal ini mengindikasikan bahwa tingkat stres remaja kemungkinan dipengaruhi oleh faktor lain di luar pilihan platform media sosial, seperti kondisi akademik, lingkungan sosial, pola tidur, maupun faktor personal lainnya.

**2. Tingkat Kecemasan Berdasarkan Interaksi Sosial**

```python
anxiety_social = df.groupby('social_interaction_level')['anxiety_level'].mean()

plt.figure(figsize=(8,5))
sns.barplot(x=anxiety_social.index, y=anxiety_social.values, palette='Blues')

plt.title('Rata-rata Tingkat Kecemasan Berdasarkan Interaksi Sosial')
plt.xlabel('Social Interaction Level')
plt.ylabel('Rata-rata Anxiety Level')
plt.show()
```

Berdasarkan visualisasi bar chart rata-rata tingkat kecemasan berdasarkan tingkat interaksi sosial, terlihat bahwa kategori high, medium, dan low memiliki rata-rata tingkat kecemasan yang relatif serupa, yaitu berada pada kisaran 5 hingga 6.

Meskipun kategori dengan interaksi sosial tinggi menunjukkan rata-rata kecemasan yang sedikit lebih besar dibandingkan kategori lainnya, perbedaannya tidak terlalu signifikan. Hal ini menunjukkan bahwa tingkat interaksi sosial dalam dataset belum menunjukkan hubungan yang kuat terhadap tingkat kecemasan remaja. Dengan demikian, diperlukan analisis lebih lanjut terhadap faktor lain yang mungkin memiliki pengaruh lebih besar terhadap kondisi kecemasan.

**3. Tingkat Kecanduan Berdasarkan Gender**

```python
addiction_gender = df.groupby('gender')['addiction_level'].mean()

plt.figure(figsize=(6,5))
sns.barplot(x=addiction_gender.index, y=addiction_gender.values, palette='Set2')

plt.title('Rata-rata Tingkat Kecanduan Berdasarkan Gender')
plt.xlabel('Gender')
plt.ylabel('Rata-rata Addiction Level')
plt.show()
```

Berdasarkan visualisasi bar chart rata-rata tingkat kecanduan media sosial berdasarkan gender, terlihat bahwa remaja laki-laki (male) dan perempuan (female) memiliki tingkat kecanduan yang hampir sama. Rata-rata tingkat kecanduan pada kedua kelompok berada pada kisaran 5 hingga 6.

Perbedaan yang sangat kecil antara kedua gender menunjukkan bahwa kecenderungan kecanduan media sosial tidak dipengaruhi secara signifikan oleh jenis kelamin. Hal ini mengindikasikan bahwa penggunaan media sosial yang intens dapat terjadi pada remaja laki-laki maupun perempuan dengan tingkat yang relatif setara. ## **Composition (Komposisi)**

**1. Komposisi Platform Media Sosial**

```python
platform_counts = df['platform_usage'].value_counts()

plt.figure(figsize=(7,6))
plt.pie(platform_counts, labels=platform_counts.index, autopct='%1.1f%%', startangle=140)

plt.title('Komposisi Penggunaan Platform Media Sosial')
plt.axis('equal')
plt.show()
```

Berdasarkan visualisasi pie chart komposisi penggunaan platform media sosial, terlihat bahwa penggunaan platform oleh remaja relatif seimbang. Pengguna Instagram memiliki proporsi sebesar 34,2%, diikuti oleh TikTok sebesar 33,2%, dan penggunaan kedua platform (Both) sebesar 32,6%.

## **Outliers Values**

Pada tahap ini, kita mengidentifikasi nilai ekstrem (outliers) pada kolom yang bertipe data numerik. Tujuan dari proses ini adalah untuk memahami seberapa besar distribusi data yang menyimpang dari pola umum, sehingga dapat meningkatkan akurasi analisis dan model yang akan dibangun.

Namun, tidak semua kolom dengan outliers perlu langsung ditangani. Kita tetap harus mempertimbangkan konteks dan makna dari setiap variabel dalam dataset.

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

# Dataframe dari list hasil
results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

# Tampilkan dataframe
display(results_df)
```

Berdasarkan hasil pengecekan outlier pada kolom depression_label, diketahui terdapat persentase outlier sebesar 2,58%.

Namun, kolom depression_label merupakan variabel kategorikal biner yang hanya memiliki dua nilai valid, yaitu:

0 = tidak depresi
1 = depresi

Sehingga nilai yang terdeteksi sebagai outlier sebenarnya bukan kesalahan data, melainkan berasal dari distribusi data yang tidak seimbang (imbalanced data), di mana jumlah responden dengan label depresi lebih sedikit dibandingkan responden yang tidak depresi.

Oleh karena itu, outlier pada kolom ini tidak perlu dihapus maupun dilakukan penanganan menggunakan metode seperti IQR atau Z-Score karena dapat menghilangkan informasi penting pada target klasifikasi.

Kesimpulan: Kolom depression_label tetap dipertahankan tanpa penanganan outlier karena merupakan variabel target klasifikasi biner dan nilai yang terdeteksi masih valid secara logika data.
